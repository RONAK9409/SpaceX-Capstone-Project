```python
import pandas as pd
import sqlite3
```

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"

df = pd.read_csv(url)

conn = sqlite3.connect("spacex.db")

df.to_sql("SPACEXTBL", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT LaunchSite,
COUNT(*) AS TotalLaunches
FROM SPACEXTBL
GROUP BY LaunchSite
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT Orbit,
COUNT(*) AS Total
FROM SPACEXTBL
GROUP BY Orbit
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT Outcome,
COUNT(*) AS Total
FROM SPACEXTBL
GROUP BY Outcome
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT AVG(PayloadMass)
AS AveragePayload
FROM SPACEXTBL
"""

pd.read_sql(query, conn)
```
