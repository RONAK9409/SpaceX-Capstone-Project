```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"

df = pd.read_csv(url)

df.head()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='LaunchSite', data=df)
plt.title('Launches by Launch Site')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Orbit', data=df)
plt.title('Orbit Distribution')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='FlightNumber', y='PayloadMass', data=df)
plt.title('Payload Mass vs Flight Number')
plt.show()
```

```python
df['Year'] = pd.to_datetime(df['Date']).dt.year

launches_per_year = df.groupby('Year').size()

launches_per_year.plot(kind='line', marker='o')
plt.title('Launches per Year')
plt.ylabel('Number of Launches')
plt.show()
```
